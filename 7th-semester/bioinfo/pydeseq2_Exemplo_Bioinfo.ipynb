```python
import numpy as np
import pandas as pd
from pydeseq2.dds import DeseqDataSet
from pydeseq2.ds import DeseqStats
```

```python
np.random.seed(42)

n_genes = 1000
n_samples = 20

# Criando matriz de contagem (inteiros)
counts = np.random.poisson(lam=50, size=(n_samples, n_genes))

# Criando DataFrame
counts_df = pd.DataFrame(
    counts,
    columns=[f"gene_{i}" for i in range(n_genes)]
)

# Criando metadados (duas condições)
conditions = ["control"] * 10 + ["treated"] * 10

metadata = pd.DataFrame({
    "condition": conditions
})

counts_df.head()
```

```python
from pydeseq2.dds import DeseqDataSet

dds = DeseqDataSet(
    counts=counts_df,
    metadata=metadata,   # <-- aqui está a correção
    design_factors="condition"
)
```

```python
dds.deseq2()
```

```python
stat_res = DeseqStats(
    dds,
    contrast=("condition", "treated", "control")
)
```

```python
stat_res.summary()
```

```python
stat_res.run_wald_test()
stat_res.summary()
```

**Essa tabela mostra quais genes mudaram, quanto mudaram e se essa mudança é estatisticamente confiável.**

```python
results_df = stat_res.results_df
results_df.head()
```
